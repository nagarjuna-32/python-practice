```python
df = pd.read_csv("/content/Student_Performance_100_Notclean.csv")
df.head(7)
```

```python
subjects_cols =["Python","	Mathematics","DBMS	","AI","DataStructures"]
pass_mark = 40
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace = True)
df.isnull().sum()
```

```python
print("number of duplicate rows:", df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)
print("number of duplicate rows:", df.duplicated().sum())
```


create columns for total average ,



```python
cols = ["Python", "Mathematics", "DBMS", "AI", "DataStructures"]
df["Total"] = df[cols].sum(axis=1)
df["Average"] = df[cols].mean(axis=1)
df.head()
```

```python
avg_of_avg = df["Average"].mean()
print("Average of Average:", round(avg_of_avg, 2))
```
